```python
# Etape 1 - Importer les librairies
import numpy as np
import pandas as pd
```

```python
# Etape 2 - Définir les données d'entrainement porte AND
# Entrées : X (2 features)
# Sorties : Y (0,1)
# Exemple : Porte AND
# ---------------------------------------------------------
# X représente les ENTRÉES (features ou caractéristiques)
# C'est un tableau à 2 dimensions (une matrice).
# Chaque ligne correspond à un cas de test possible, 
# et chaque colonne représente une entrée (ici x₁ et x₂).
# ---------------------------------------------------------
X = np.array([
    [0, 0],  # Cas 1 : l'entrée 1 vaut 0, l'entrée 2 vaut 0
    [0, 1],  # Cas 2 : l'entrée 1 vaut 0, l'entrée 2 vaut 1
    [1, 0],  # Cas 3 : l'entrée 1 vaut 1, l'entrée 2 vaut 0
    [1, 1]   # Cas 4 : l'entrée 1 vaut 1, l'entrée 2 vaut 1
])

# ---------------------------------------------------------
# y représente les SORTIES ATTENDUES (les étiquettes ou "targets")
# C'est un tableau à 1 dimension (un vecteur).
# Chaque valeur correspond au résultat souhaité pour la ligne de X au même indice.
# Ici, on simule la porte logique AND (ET) : 
# La sortie vaut 1 UNIQUEMENT si les deux entrées valent 1.
# ---------------------------------------------------------
y = np.array([0, 0, 0, 1])

# Visualisation sous forme de DataFrame
df_X = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])
display(df_X)

# Création d'un dataframe pour y
df_y = pd.DataFrame(y, columns=['Target'])
display(df_y)
```

```python
# Etape 3 - Initialiser les poids et le biais

# =========================================================
# RAPPEL THÉORIQUE : C'est quoi un Poids et un Biais ?
# - Les poids (w) : Représentent l'importance (l'influence) de chaque entrée.
#   Un grand poids pousse le neurone à dire "oui" (1), un poids négatif le pousse à dire "non" (0).
# - Le biais (b) : Représente la tendance naturelle ou le seuil d'activation du neurone 
#   (comme un niveau de optimisme/pessimisme de base avant même de regarder les entrées).
# =========================================================

# Fixe la graine aléatoire (seed) pour garantir la reproductibilité : 
# à chaque fois que je relancerai le code, NumPy générera exactement les mêmes nombres aléatoires.
np.random.seed(42)

# W : Initialise un tableau de 2 poids aléatoires (un pour chaque entrée/feature)
w = np.random.randn(2)

# b : Initialise un biais (seuil) aléatoire unique (scalaire)
b = np.random.randn()


# Affiche les valeurs initiales pour vérification 
print("Poids initiaux:", w)
print("Biais initial:", b)
```

```python
# =========================================================
# ÉTAPE 4 : Définir la fonction d'activation (la Sigmoïde)
# =========================================================
# Rôle : Elle transforme n'importe quel nombre réel (qui peut aller de moins l'infini à plus l'infini)
# en une valeur strictement comprise entre 0 et 1. 
# Pourquoi ? Pour que le neurone puisse interpréter sa sortie comme une PROBABILITÉ 
# (par exemple : 0.9 = 90% de chances que ce soit vrai, 0.1 = 10% de chances).
# Sa courbe en forme de "S" aplati permet aussi d'introduire de la non-linéarité.
# =========================================================

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
# =========================================================
# ÉTAPE 5 : Définir la prédiction du perceptron (Forward propagation)
# =========================================================
# Rôle : C'est le moment où le neurone "réfléchi" en deux temps :
# 1. z : La somme pondérée (on multiplie chaque entrée par son poids, et on ajoute le biais).
# 2. On passe ce résultat z dans la sigmoïde pour obtenir une probabilité finale entre 0 et 1.
# =========================================================

def predict(X, w, b):
    # np.dot(X, w) réalise le produit matriciel (la somme pondérée pour toutes les lignes d'un coup)
    # On y ajoute le biais (b) pour obtenir le score net (z)
    z = np.dot(X, w) + b  
    
    # On applique la fonction sigmoïde pour transformer ce score en probabilité
    return sigmoid(z)
```

```python
# ---------------------------------------------------------
# ÉTAPE 6 : Definir la fonction de perte (Loss) - Erreur Quadratique Moyenne (MSE)
# Elle mesure l'écart global entre ce que le modèle prédit (y_pred) 
# et la réalité (y_true). Plus l'erreur est proche de 0, meilleur est le modèle.
# ---------------------------------------------------------
def loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

```python
# =========================================================
# ÉTAPE 7 : Entraînement du perceptron (descente de gradient)
# =========================================================
learning_rate = 0.1
epochs = 2000

for i in range(epochs):
    y_pred = predict(X, w, b)
    
    # Calcul du gradient (avec la formule exacte du support incluant le -2)
    dw = -2 * np.dot((y - y_pred) * y_pred * (1 - y_pred), X) / len(X)
    db = -2 * np.mean((y - y_pred) * y_pred * (1 - y_pred))
    
    # Mise à jour des poids et du biais
    w -= learning_rate * dw
    b -= learning_rate * db
    
    # Affichage de l'erreur toutes les 100 époques (comme dans l'énoncé)
    if i % 100 == 0:
        print(f"Epoch {i}, Loss: {loss(y, y_pred)}")
```

```python
### Étape 8 : Tester le perceptron AND

# Demande au neurone entraîné de faire ses prédictions sur les entrées X
y_pred = predict(X, w, b)

print("Prédictions finales (probabilités brutes) :", y_pred)

# Convertit les probabilités en classes binaires strictes (0 ou 1) en utilisant un seuil de 0.5
print("Classes prédites (seuil 0.5) :", (y_pred > 0.5).astype(int))
```

```python
# =========================================================
# POUR ALLER PLUS LOIN : L'entropie croisée binaire (BCE)
# =========================================================
# Réinitialisation des poids de manière aléatoire

np.random.seed(42)
w = np.random.randn(2)
b = np.random.randn()

# Définition de la fonction de perte "Entropie Croisée"
def bce(y_true, y_pred, eps=1e-12):
    # eps évite d'avoir un log(0) qui planterait le programme (division par zéro)
    return -np.mean(y_true * np.log(y_pred + eps) + (1 - y_true) * np.log(1 - y_pred + eps))

# Un learning rate un peu plus élevé (0.5) car le gradient est plus direct
learning_rate = 0.5  

for i in range(2000):
    y_pred = predict(X, w, b)
    
    # Gradient simplifié de l'entropie croisée couplée à la sigmoïde : dz = y_pred - y
    dz = y_pred - y  
    
    # Mise à jour des poids et du biais
    w -= learning_rate * np.dot(X.T, dz) / len(X)
    b -= learning_rate * np.mean(dz)
    
    # Affichage de l'évolution de l'erreur toutes les 400 époques
    if i % 400 == 0:
        print(f"Epoch {i}, Loss: {bce(y, y_pred):.4f}")

print("Prédictions finales (BCE) :", predict(X, w, b).round(3))
```
